```python
import numpy as np
from sklearn.datasets import load_iris
```

```python
data = load_iris()

X = data.data
y = data.target

print(X.shape)
```

```python
np.random.seed(42)

m = X.shape[0]
indices = np.random.permutation(m)

split_index = int(0.8*m)

train_indices = indices[:split_index]
test_indices = indices[split_index:]

X_train = X[train_indices]
y_train = y[train_indices]
X_test = X[test_indices]
y_test = y[test_indices]
```

```python
mean = np.mean(X_train,axis=0)
std = np.std(X_train,axis=0)

X_train = (X_train - mean)/std
X_test = (X_test-mean)/std
```

```python
def euclidean_distance(x1,x2):
  return np.sqrt(np.sum((x1 - x2)**2))
```

```python
def predict_one(X_train, y_train, x, k=5):
  distances = []
  for i in range(len(X_train)):
    distance = euclidean_distance(X_train[i], x)
    distances.append(distance)

  distances = np.array(distances)
  #[3,2,5]
  #[2,3,5]
  #[1,0,2]
  sorted_distances = np.argsort(distances)
  k_nearest_indices = sorted_distances[:k]
  k_nearest_labels = y_train[k_nearest_indices]
  counts = np.bincount(k_nearest_labels)
  return np.argmax(counts)
```

```python
def predict(X_train,y_train,X_test,k):
  predictions = []
  for x in X_test:
    prediction = predict_one(X_train,y_train,x,k)
    predictions.append(prediction)
  return np.array(predictions)
```

```python
k = 5
predictions = predict(X_train,y_train,X_test,k)
```

```python
#[1,0,1,1]
#[1,1,1,1]
#0.75
accuracy = np.mean(predictions == y_test)*100

print(accuracy)
```
